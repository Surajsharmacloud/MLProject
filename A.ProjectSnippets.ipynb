# Project Summary

This project is an end-to-end Machine Learning application that predicts a student's Mathematics score based on demographic and academic features. The workflow begins with data ingestion, followed by preprocessing through pipelines, model training, evaluation, and serialization of both the preprocessor and the trained model. A Flask web application collects user inputs, transforms them using the saved preprocessor, generates predictions using the saved model, and displays the result on a web page. The application was deployed to both AWS Elastic Beanstalk and Azure App Service, with automated CI/CD implemented using GitHub Actions.
